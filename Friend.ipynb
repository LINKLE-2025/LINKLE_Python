```python
import pymysql
import random

# ---------------------------
# DB 연결 설정
# ---------------------------
conn = pymysql.connect(
    host="localhost",     # DB 주소
    port=3333,            # MariaDB 포트
    user="linkleTest",    # 사용자 이름
    password="1111",      # 비밀번호
    database="linkleTest2",   # 사용할 DB 스키마명
    charset="utf8mb4"
)

cursor = conn.cursor()

# ---------------------------
# 랜덤 FRIEND 데이터 생성
# ---------------------------
def insert_random_friends(total_users=20322, num_records=1000):
    inserted_count = 0

    for _ in range(num_records):
        user1 = random.randint(1, total_users)
        user2 = random.randint(1, total_users)

        # 자기 자신은 친구 불가
        if user1 == user2:
            continue

        # 항상 user1 < user2 로 정렬해서 양방향 중복 방지
        if user1 > user2:
            user1, user2 = user2, user1

        # 상태 무작위 (REQUESTED, ACCEPTED)
        state = random.choice(["REQUESTED", "ACCEPTED"])

        try:
            sql = """
                INSERT INTO FRIEND (user_id1, user_id2, state)
                VALUES (%s, %s, %s)
            """
            cursor.execute(sql, (user1, user2, state))
            inserted_count += 1

        except pymysql.err.IntegrityError:
            # 이미 같은 (user1, user2) 조합 존재 시 무시
            continue

    conn.commit()
    print(f"{num_records}개 시도 → {inserted_count}개 추가 완료")

insert_random_friends(num_records=38765)
    
```
